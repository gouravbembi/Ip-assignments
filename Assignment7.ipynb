```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# -----------------------------------------
# Step 1: Upload Input Image
# -----------------------------------------

uploaded = files.upload()

filename = list(uploaded.keys())[0]

img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image could not be loaded!")

# -----------------------------------------
# 1. LAPLACIAN FILTER
# -----------------------------------------

# Apply Laplacian filter
laplacian = cv2.Laplacian(img, cv2.CV_64F)

# Convert Laplacian result to uint8
laplacian = np.uint8(np.absolute(laplacian))

# Sharpen the image by subtracting Laplacian
laplacian_sharp = cv2.subtract(img, laplacian)

# -----------------------------------------
# 2. SOBEL GRADIENT OPERATOR
# -----------------------------------------

# Apply Sobel operator in X direction
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)

# Apply Sobel operator in Y direction
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Calculate gradient magnitude
sobel_magnitude = np.sqrt(sobel_x**2 + sobel_y**2)

# Normalize Sobel result
sobel_magnitude = cv2.normalize(
    sobel_magnitude, None, 0, 255, cv2.NORM_MINMAX
).astype(np.uint8)

# Sharpen image using Sobel gradient
sobel_sharp = cv2.add(img, sobel_magnitude)

# -----------------------------------------
# CANNY EDGE DETECTOR
# -----------------------------------------

canny = cv2.Canny(img, 100, 200)

# Sharpen image using Canny edges
canny_sharp = cv2.add(img, canny)

# -----------------------------------------
# 3. UNSHARP MASKING
# -----------------------------------------

# Gaussian filter of size 5x5 and sigma = 3
blurred = cv2.GaussianBlur(
    img, (5, 5), sigmaX=3, sigmaY=3
)

# Create mask by subtracting blurred image
mask = cv2.subtract(img, blurred)

# Add mask to original image
unsharp = cv2.add(img, mask)

# -----------------------------------------
# Step 4: Display All Results
# -----------------------------------------

plt.figure(figsize=(15, 12))

plt.subplot(3, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(3, 3, 2)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Filter")
plt.axis("off")

plt.subplot(3, 3, 3)
plt.imshow(laplacian_sharp, cmap='gray')
plt.title("Laplacian Sharpened")
plt.axis("off")

plt.subplot(3, 3, 4)
plt.imshow(sobel_magnitude, cmap='gray')
plt.title("Sobel Gradient")
plt.axis("off")

plt.subplot(3, 3, 5)
plt.imshow(sobel_sharp, cmap='gray')
plt.title("Sobel Sharpened")
plt.axis("off")

plt.subplot(3, 3, 6)
plt.imshow(canny, cmap='gray')
plt.title("Canny Edge Detector")
plt.axis("off")

plt.subplot(3, 3, 7)
plt.imshow(canny_sharp, cmap='gray')
plt.title("Canny Sharpened")
plt.axis("off")

plt.subplot(3, 3, 8)
plt.imshow(blurred, cmap='gray')
plt.title("Gaussian Blurred (5x5, sigma=3)")
plt.axis("off")

plt.subplot(3, 3, 9)
plt.imshow(unsharp, cmap='gray')
plt.title("Unsharp Masking")
plt.axis("off")

plt.tight_layout()
plt.show()
```
