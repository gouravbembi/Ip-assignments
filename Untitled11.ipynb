```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))
image = cv2.imread(filename)

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

R = image[:, :, 0]
G = image[:, :, 1]
B = image[:, :, 2]

gray_mean = ((R.astype(float) + G.astype(float) + B.astype(float)) / 3).astype(np.uint8)

wr = float(input("Enter weightage of R: "))
wg = float(input("Enter weightage of G: "))
wb = float(input("Enter weightage of B: "))

if wr < 0 or wr > 1 or wg < 0 or wg > 1 or wb < 0 or wb > 1 or abs(wr + wg + wb - 1) > 0.0001:
    print("Invalid weightages")
else:
    gray_weighted = (wr * R + wg * G + wb * B).astype(np.uint8)

    plt.figure(figsize=(12, 8))

    plt.subplot(2, 2, 1)
    plt.imshow(image)
    plt.title("RGB Image")
    plt.axis("off")

    plt.subplot(2, 2, 2)
    plt.imshow(gray_mean, cmap="gray")
    plt.title("Grayscale - Mean Average")
    plt.axis("off")

    plt.subplot(2, 2, 3)
    plt.imshow(image)
    plt.title("RGB Image")
    plt.axis("off")

    plt.subplot(2, 2, 4)
    plt.imshow(gray_weighted, cmap="gray")
    plt.title("Grayscale - Weighted Average")
    plt.axis("off")

    plt.show()
```
