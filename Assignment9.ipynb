```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from math import log10

uploaded = files.upload()
filename = list(uploaded.keys())[0]

img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise ValueError("Image could not be loaded!")

Smin = int(input("Enter initial window size (odd number): "))
Smax = int(input("Enter maximum window size (odd number): "))

if Smin < 3 or Smin % 2 == 0 or Smax % 2 == 0 or Smax < Smin:
    raise ValueError("Enter valid odd window sizes with Smax >= Smin.")

noise_img = img.copy()

prob = 0.05
rand = np.random.rand(*img.shape)

noise_img[rand < prob / 2] = 0
noise_img[(rand >= prob / 2) & (rand < prob)] = 255

median_img = cv2.medianBlur(noise_img, Smin)

def adaptive_median(image, Smin, Smax):

    padded = cv2.copyMakeBorder(
        image, Smax//2, Smax//2,
        Smax//2, Smax//2,
        cv2.BORDER_REFLECT
    )

    result = image.copy()
    rows, cols = image.shape

    for i in range(rows):
        for j in range(cols):

            size = Smin
            output = image[i, j]

            while size <= Smax:

                r = Smax//2 + i
                c = Smax//2 + j
                k = size//2

                window = padded[
                    r-k:r+k+1,
                    c-k:c+k+1
                ]

                zmin = np.min(window)
                zmax = np.max(window)
                zmed = np.median(window)
                zxy = image[i, j]

                A1 = zmed - zmin
                A2 = zmed - zmax

                if A1 > 0 and A2 < 0:

                    B1 = zxy - zmin
                    B2 = zxy - zmax

                    if B1 > 0 and B2 < 0:
                        output = zxy
                    else:
                        output = zmed

                    break

                else:
                    size += 2

            if size > Smax:
                output = zmed

            result[i, j] = output

    return result.astype(np.uint8)

adaptive_img = adaptive_median(noise_img, Smin, Smax)

def calculate_psnr(original, denoised):

    mse = np.mean(
        (original.astype(np.float64) -
         denoised.astype(np.float64)) ** 2
    )

    if mse == 0:
        return float('inf')

    return 10 * log10((255 ** 2) / mse)

psnr_noisy = calculate_psnr(img, noise_img)
psnr_median = calculate_psnr(img, median_img)
psnr_adaptive = calculate_psnr(img, adaptive_img)

images = [
    img,
    noise_img,
    median_img,
    adaptive_img
]

titles = [
    "Original Image",
    "Salt and Pepper Noise",
    "Median Filter",
    "Adaptive Median Filter"
]

plt.figure(figsize=(12, 8))

for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()

print("PSNR Comparison")
print("-------------------------")
print("Noisy Image PSNR:", psnr_noisy)
print("Median Filter PSNR:", psnr_median)
print("Adaptive Median PSNR:", psnr_adaptive)

plt.figure(figsize=(7, 4))

plt.bar(
    ["Noisy", "Median", "Adaptive Median"],
    [psnr_noisy, psnr_median, psnr_adaptive]
)

plt.ylabel("PSNR (dB)")
plt.title("PSNR Comparison")
plt.show()
```
